```python
# Find-S Algorithm

# Dataset
# Last column is the target/class
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'No'],
    ['Sunny', 'Warm', 'High', 'Weak', 'Yes']
]

# Start with the most specific hypothesis
hypothesis = ['Ø', 'Ø', 'Ø', 'Ø']

# Process each example
for row in data:

    # Check whether the example is positive
    if row[-1] == 'Yes':

        # If this is the first positive example
        if hypothesis[0] == 'Ø':
            hypothesis = row[:-1]

        else:
            # Compare each attribute
            for i in range(len(hypothesis)):

                # If values are different, make it '?'
                if hypothesis[i] != row[i]:
                    hypothesis[i] = '?'

# Display final hypothesis
print("Final Hypothesis:")
print(hypothesis)
```
